```python
# 1.Even or Odd Function
# Write a function is_even(n) that returns "Even" if the number is even, otherwise "Odd".
n = int(input("Enter the Number: "))
def is_even(n):
    if(n%2==0):
        return "Even"
    else:
        return "Odd"

print(is_even(n))
```

```python
# 2.Largest of Two Numbers
# Write a function largest(a, b) that returns the larger number
a= int(input("Enter the Number: "))
b= int(input("Enter the Number: "))
def largest(a, b):
    if(a != b):
        if(a>b):
            return a
        else:
            return b

    else:
        return "Both are equal"

print(largest(a,b))
```

```python
#Write a function that prints numbers from 1 to 100.If divisible by 3, print "Fizz".If divisible by 5, print "Buzz".If divisible by both, print "FizzBuzz".
def fizzbuzz():
    for i in range(1, 101):

        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")

        elif i % 3 == 0:
            print("Fizz")

        elif i % 5 == 0:
            print("Buzz")

        else:
            print(i)

fizzbuzz()
```

```python
#Write a function that returns True if a string is a palindrome.
def palindrome(s):
    if s == s[::-1]:
        return True
    else:
        return False

print(palindrome("madam"))
print(palindrome("hello"))
```

```python
#Write a function that counts the vowels in a string.

def count_vowels(s):
    count = 0

    for i in s:
        if i.lower() in "aeiou":
            count += 1

    return count

print(count_vowels("Education"))
```

```python
#Write a function that returns the second largest number in a list without using sort().
def second_largest(numbers):
    largest = float('-inf')
    second = float('-inf')

    for i in numbers:

        if i > largest:
            second = largest
            largest = i

        elif largest > i > second:
            second = i

    if second == float('-inf'):
        return None

    return second

print(second_largest([10, 20, 40, 30, 50]))
```

```python
'''Generate a random number between 1 and 10.
Keep asking the user to guess until they get it right.
Use a loop, conditionals, and functions.'''

import random

def guessing_game():
    number = random.randint(1, 10)

    while True:
        guess = int(input("Guess a number between 1 and 10: "))

        if guess == number:
            print("Correct Guess!")
            break

        elif guess < number:
            print("Too Low!")

        else:
            print("Too High!")

guessing_game()
```

```python
'''Write a function grade(marks) that:
Returns "A" for marks ≥ 90
Returns "B" for marks ≥ 80
Returns "C" for marks ≥ 70
Returns "D" for marks ≥ 60
Otherwise returns "F"
'''

def grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    elif marks >= 60:
        return "D"

    else:
        return "F"

print(grade(95))
print(grade(85))
print(grade(55))
```
